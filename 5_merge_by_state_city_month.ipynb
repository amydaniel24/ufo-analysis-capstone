```python
import pandas as pd

ufo_df = pd.read_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\cleaned_ufo_data.csv')
drone_df = pd.read_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\cleaned_drone_data.csv')
parks_df = pd.read_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\national_parks_clean.csv')
density_df = pd.read_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\us_city_population_density_clean.csv')
```

```python
for df in [ufo_df, drone_df, parks_df, density_df]:
    df['state'] = df['state'].str.upper()
    df['city'] = df['city'].str.upper()
```

```python
for df in [ufo_df, drone_df, parks_df]:
    if 'datetime' in df.columns:
        df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
        df['month'] = df['datetime'].dt.month
```

```python
print(ufo_df.columns.tolist())
print(drone_df.columns.tolist())
print(parks_df.columns.tolist())
```

```python
ufo_df['datetime'] = pd.to_datetime(ufo_df['datetime'], errors='coerce')
ufo_df['month'] = ufo_df['datetime'].dt.month

drone_df['date'] = pd.to_datetime(drone_df['date'], errors='coerce')
drone_df['month'] = drone_df['date'].dt.month

parks_df['datetime'] = pd.to_datetime(parks_df['datetime'], errors='coerce')
parks_df['month'] = parks_df['datetime'].dt.month
```

```python
ufo_drone_df = pd.merge(
    ufo_df, drone_df,
    on=['state', 'city', 'month'], 
    how='outer', 
    suffixes=('_ufo', '_drone',)
)
```

```python
ufo_drone_parks_df = pd.merge(
    ufo_drone_df,
    parks_df,
    on=['state', 'city', 'month'],
    how='outer',
    suffixes=('', '_parks'))
```

```python
final_df = pd.merge(
    ufo_drone_parks_df, density_df, 
    on=['state', 'city'], 
    how='left'
)
```

```python
final_df.to_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\merged_ufo_drone_parks_density.csv', index=False)
final_df.head()
```
