```python
import pandas as pd

density_df = pd.read_csv(r'C:\Users\opiej\ufo-analysis-capstone\ufo-analysis-capstone\data\us_city_population_density.csv')
```

```python
print(density_df.head())
print(density_df.columns.tolist())
```

```python
density_df.columns = density_df.columns.str.strip().str.lower().str.replace('', '')
print(density_df.columns.tolist())
```

```python
density_df.columns = (
    density_df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('(', '')
    .str.replace(')', '')
    .str.replace('/', '_')
)
```

```python
density_df = density_df.rename(columns={
    'population_density_persons_square_mile': 'population_density',
    '2016_population': 'population',
    'land_area_square_miles': 'land_area'
})
```

```python
print(density_df[['city', 'state', 'population_density', 'population']].head())
```

```python
print("Missing:", density_df.isna().sum())
print("Duplicates:", density_df.duplicated().sum())
```

```python
top10 = density_df.sort_values(by='population_density', ascending=False).head(10)
print(top10[['city','state','population_density']])
```
